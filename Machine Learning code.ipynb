```python
import numpy as np
import pandas as pd
```

```python
movies = pd.read_csv('tmdb_5000_movies.csv')
credits = pd.read_csv('tmdb_5000_credits.csv')
```

```python
movies.head()
```

```python
credits.head()
```

```python
movies = movies.merge(credits, on='title')
```

```python
movies.shape
```

```python
movies.info()
```

```python
#We will use and keep only following columns
#genres
#id
#keywords
#title
#overview
#cast
#crew
movies = movies[['genres','id','keywords','title','overview','cast','crew']]
```

```python
movies.head()
```

```python
movies.isnull().sum()
```

```python
movies.dropna(inplace=True)
```

```python
movies.duplicated().sum()
```

```python
import ast
```

```python
def convert(obj):
    L=[]
    for i in ast.literal_eval(obj):
       L.append( i['name'])
    return L
```

```python
movies['genres'] = movies['genres'].apply(convert)
```

```python
movies['keywords'] = movies['keywords'].apply(convert)
```

```python
def convert3(obj):
    L=[]
    counter = 0
    for i in ast.literal_eval(obj):
        if counter !=3:
           L.append( i['name'])
           counter= counter + 1
        else:
            break
    return L
```

```python
movies['cast'] = movies['cast'].apply(convert3)
```

```python
def fetch_director(obj):
    L=[]
    for i in ast.literal_eval(obj):
      if i['job']== 'Director':
       L.append( i['name'])
       break
    return L
```

```python
movies['crew'] = movies['crew'].apply(fetch_director)
```

```python
movies.head()
```

```python
movies['overview'] = movies['overview'].apply(lambda x: x.split())
```

```python
movies['genres'] = movies['genres'].apply(lambda x: [i.replace(" ","") for i in x])
```

```python
movies['keywords'] = movies['keywords'].apply(lambda x: [i.replace(" ","") for i in x])
```

```python
movies['cast'] = movies['cast'].apply(lambda x: [i.replace(" ","") for i in x])
```

```python
movies['crew']= movies['crew'].apply(lambda x: [i.replace(" ","") for i in x])
```

```python
movies['tags'] =movies['overview'] + movies['genres']+movies['keywords']+ movies['cast']+ movies['crew']
```

```python
new_df = movies[['id','title','tags']]
```

```python
new_df['tags']= new_df['tags'].apply(lambda x : " ". join(x))
```

```python
new_df.head()
```

```python
new_df['tags'][0]
```

```python
new_df['tags']= new_df['tags'].apply(lambda x: x.lower())
```

```python
new_df.head()
```

```python
import nltk
```

```python
from nltk.stem.porter import PorterStemmer
ps= PorterStemmer()
```

```python
def stem (text):
    L=[]                       #this will convert words like dance, dancing , danced to danc
    for i in text.split():
      L.append(ps.stem(i))
    return " ".join(L)
```

```python
new_df['tags'] = new_df['tags'].apply(stem)
```

```python
import sklearn
```

```python
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer(max_features=5000, stop_words= 'english') #stop words(in,the,are) words will be dropped and rest 5000 words will become vectors out of all tags
```

```python
vectors= cv.fit_transform(new_df['tags']).toarray()
```

```python
cv.get_feature_names_out()
```

```python
from sklearn.metrics.pairwise import cosine_similarity #now calculating similarity b/w vectors using cosine distance
```

```python
similarity= cosine_similarity(vectors)
```

```python
def recommend(movie):
    movie_index=new_df[new_df['title']== movie].index[0]
    distance= similarity[movie_index]
    movie_list= sorted(list(enumerate(distance)),reverse=True, key= lambda x:x[1]) [1:6]
    for i in movie_list:
        print(new_df.iloc[i[0]].title)
```

```python
recommend('Batman Begins')
```
